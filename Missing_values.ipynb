```python
import pandas as pd
import numpy as np
```

```python
students = pd.DataFrame({
    "Student_Id": [101,102,102,103,104,105],
    "Name":["Rashi","Yashi","Rahul","Yashi ","Yug",None],
    "Department":["CSE","cse","cse","Data Science","DATA SCIENCE","ECE"],
    "Marks":[85,72,72,np.nan,91,67],
    "Attendance":[90,82,82,75,np.nan,88],
    "Exam_Date":["2026-09-01","01/09/2026","01/09/2026","2026-09-02","not known","2026=09-03"]
})
```

# **Find Missing Values**

```python
print(students.isna())
print("\nMissing count in each column:")
students.isna().sum()
```

# **Remove rows containing any missing value**

```python
df_clean=students.dropna()
df_clean
```

# **Remove columns containing any missing value**

```python
df_clean=students.dropna(axis=1)
df_clean
```

# **Remove rows only when all values are missing**

```python
df_clean=students.dropna(how="all")
df_clean
```

# **Replace missing values using mean**

```python
clean=students.copy()
clean["Marks"]=clean["Marks"].fillna(clean["Marks"].mean())
clean["Attendance"]=clean["Attendance"].fillna(clean["Attendance"].mean())
clean["Name"]=clean["Name"].fillna("Unknown")
print(clean[["Name","Marks","Attendance"]])
```

```python
clean=students.copy()
clean["Marks"]=clean["Marks"].fillna(clean["Marks"].median())
clean["Attendance"]=clean["Attendance"].fillna(clean["Attendance"].median())
clean["Name"]=clean["Name"].fillna("Unknown")
print(clean[["Name","Marks","Attendance"]])
```

```python
df_mode=students.copy()
df_mode["Name"]=df_mode["Name"].fillna(df_mode["Name"].mode()[0])
```

# **Handling missing values:SimpleImputer and KNNImputer**

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
df=pd.DataFrame({
    "Age":[25,30,np.nan,22,35],
    "Salary":[50000,60000,55000,np.nan,70000]
})
print("Original Dataframe:")
print(df)

imputer=SimpleImputer(strategy="mean")
result=pd.DataFrame(imputer.fit_transform(df),columns=df.columns)
print("\nDataframe after imputation")
print(result)
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
df=pd.DataFrame({
    "Age":[25,30,np.nan,22,35],
    "Salary":[50000,60000,55000,np.nan,70000]
})
print("Original Dataframe:")
print(df)

imputer=SimpleImputer(strategy="most_frequent")
result=pd.DataFrame(imputer.fit_transform(df),columns=df.columns)
print("\nDataframe after imputation")
print(result)
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
df=pd.DataFrame({
    "Age":[25,30,np.nan,22,35],
    "Salary":[50000,60000,55000,np.nan,70000]
})
print("Original Dataframe:")
print(df)

imputer=SimpleImputer(strategy="constant",fill_value=0)
result=pd.DataFrame(imputer.fit_transform(df),columns=df.columns)
print("\nDataframe after imputation")
print(result)
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
df=pd.DataFrame({
    "Age":[25,30,np.nan,22,35],
    "Salary":[50000,60000,55000,np.nan,70000]
})
print("Original Dataframe:")
print(df)

imputer=KNNImputer(n_neighbors=2)
result=pd.DataFrame(imputer.fit_transform(df),columns=df.columns)
print("\nDataframe after imputation")
print(result)
```
